```python
data['Age'].fillna(data['Age'].median(), inplace=True)
```

```python
data['Deck'] = data['Cabin'].str[0]
data['Deck'].fillna('Unknown', inplace=True)
data.drop('Cabin', axis=1, inplace=True)
```

```python
data['Embarked'].fillna(data['Embarked'].mode()[0], inplace=True)
```

```python
data.isna().sum()
```

```python
data.duplicated().sum()
```

```python
# Convert categorical columns
sns.set(style="whitegrid")
data['Sex'] = data['Sex'].astype('category')
data['Embarked'] = data['Embarked'].astype('category')
```

```python
# Pairplot for numeric features
sns.pairplot(data[['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']], hue='Survived')
plt.suptitle('Pairplot of Titanic Features', y=1.02)
plt.tight_layout()
plt.savefig('pairplot.png')
```

```python
# Heatmap of correlation
plt.figure(figsize=(6, 6))
sns.heatmap(data[['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare']].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.tight_layout()
plt.savefig('heatmap.png')
```

```python
# Histogram of Age
sns.histplot(data['Age'].dropna(), bins=30, kde=True)
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.tight_layout()
plt.savefig('age_histogram.png')
```

```python
# Boxplot of Age by Survival
sns.boxplot(data=data, x='Survived', y='Age')
plt.title('Age Distribution by Survival')
plt.xlabel('Survived')
plt.ylabel('Age')
plt.tight_layout()
plt.savefig('age_boxplot.png')
```

```python
# Scatterplot Fare vs Age by Survival
sns.scatterplot(data=data, x='Age', y='Fare', hue='Survived')
plt.title('Fare vs Age by Survival')
plt.tight_layout()
plt.savefig('fare_age_scatter.png')
```

```python
# Countplot of Class vs Survival
sns.countplot(data=data, x='Pclass', hue='Survived')
plt.title('Passenger Class vs Survival')
plt.tight_layout()
plt.savefig('pclass_survival.png')
```

```python
# Countplot of Embarked vs Survival
sns.countplot(data=data, x='Embarked', hue='Survived')
plt.title('Port of Embarkation vs Survival')
plt.tight_layout()
plt.savefig('embarked_survival.png')
```
